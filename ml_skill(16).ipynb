```python
# ============================================================
# EXPERIMENT
# PCA + K-MEANS CLUSTERING
#
# UNIVERSAL CSV VERSION
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


# ============================================================
# 1. UPLOAD CSV
# ============================================================

print("=" * 75)
print("UNIVERSAL PCA + K-MEANS CLUSTERING")
print("=" * 75)

print("\nUpload your CSV dataset.")

uploaded = files.upload()

if len(uploaded) == 0:
    raise ValueError("No CSV file was uploaded.")

file_path = list(uploaded.keys())[0]

print("\nSelected file:")
print(file_path)


# ============================================================
# 2. LOAD DATASET
# ============================================================

try:

    data = pd.read_csv(file_path)

except Exception as e:

    raise ValueError(
        f"Could not read the CSV file:\n{e}"
    )


print("\nDataset loaded successfully!")

print("\nShape:")
print(data.shape)

print("\nFirst 5 rows:")
print(data.head())

print("\nColumns:")
print(data.columns.tolist())


# ============================================================
# 3. BASIC CLEANING
# ============================================================

# Remove completely empty columns

empty_columns = [

    col

    for col in data.columns

    if data[col].isna().all()

]


if empty_columns:

    print(
        "\nRemoving completely empty columns:"
    )

    for col in empty_columns:

        print(" -", col)

    data = data.drop(
        columns=empty_columns
    )


# Remove duplicate rows

duplicate_count = data.duplicated().sum()


if duplicate_count > 0:

    print(
        f"\nRemoving {duplicate_count} duplicate rows."
    )

    data = data.drop_duplicates()


data = data.reset_index(drop=True)


# ============================================================
# 4. FIND NUMERIC FEATURES
# ============================================================

numeric_features = (

    data

    .select_dtypes(
        include=[
            "int64",
            "int32",
            "int16",
            "int8",

            "float64",
            "float32",
            "float16"
        ]
    )

    .columns

    .tolist()
)


print("\n" + "=" * 75)
print("NUMERIC FEATURE DETECTION")
print("=" * 75)


print(
    "\nNumeric columns found:",
    len(numeric_features)
)


print(
    numeric_features
)


# ============================================================
# 5. REMOVE ID-LIKE COLUMNS
# ============================================================

id_columns = []


for col in numeric_features:

    name = (

        str(col)
        .strip()
        .lower()
        .replace(" ", "_")
        .replace("-", "_")

    )


    unique_count = data[col].nunique(
        dropna=True
    )


    # Common ID names

    if (

        name == "id"

        or name.endswith("_id")

        or name.startswith("id_")

        or name in [

            "studentid",
            "student_id",

            "customerid",
            "customer_id",

            "userid",
            "user_id",

            "transactionid",
            "transaction_id",

            "recordid",
            "record_id"
        ]

    ):

        id_columns.append(col)


    # A numeric column with almost every
    # value unique can also be an ID

    elif (

        len(data) > 20

        and unique_count
        > 0.95 * len(data)

    ):

        id_columns.append(col)


if id_columns:

    print(
        "\nRemoving ID-like columns:"
    )

    for col in id_columns:

        print(
            " -",
            col
        )

    numeric_features = [

        col

        for col in numeric_features

        if col not in id_columns

    ]


# ============================================================
# 6. REMOVE CONSTANT FEATURES
# ============================================================

constant_columns = []


for col in numeric_features:

    if data[col].nunique(
        dropna=True
    ) <= 1:

        constant_columns.append(col)


if constant_columns:

    print(
        "\nRemoving constant columns:"
    )

    for col in constant_columns:

        print(
            " -",
            col
        )

    numeric_features = [

        col

        for col in numeric_features

        if col not in constant_columns

    ]


# ============================================================
# 7. CHECK FEATURES
# ============================================================

if len(numeric_features) < 2:

    raise ValueError(

        "\nAt least 2 usable numeric features "
        "are required for PCA and clustering."

    )


print(
    "\nFinal features used:"
)

print(
    numeric_features
)


# ============================================================
# 8. CREATE NUMERIC DATA
# ============================================================

X = data[
    numeric_features
].copy()


# ============================================================
# 9. HANDLE MISSING VALUES
# ============================================================

missing_values = X.isna().sum()


if missing_values.sum() > 0:

    print(
        "\nMissing values detected."
    )

    print(
        missing_values[
            missing_values > 0
        ]
    )


    imputer = SimpleImputer(
        strategy="median"
    )


    X_clean = imputer.fit_transform(
        X
    )


else:

    X_clean = X.values


# ============================================================
# 10. DATASET DESCRIPTION
# ============================================================

print("\n" + "=" * 75)
print("DATA DESCRIPTION")
print("=" * 75)


print(
    "\nDescriptive statistics:"
)


print(
    X.describe()
    .round(2)
)


# ============================================================
# 11. CORRELATION HEATMAP
# ============================================================

if len(numeric_features) <= 30:

    plt.figure(
        figsize=(10, 7)
    )


    correlation = X.corr()


    sns.heatmap(

        correlation,

        annot=True,

        fmt=".2f",

        cmap="coolwarm",

        center=0,

        square=False

    )


    plt.title(
        "Feature Correlation Heatmap"
    )


    plt.tight_layout()

    plt.show()


else:

    print(
        "\nToo many features for an annotated "
        "heatmap. Heatmap skipped."
    )


# ============================================================
# 12. STANDARDIZATION
# ============================================================

print("\n" + "=" * 75)
print("STANDARDIZATION")
print("=" * 75)


scaler = StandardScaler()


X_scaled = scaler.fit_transform(
    X_clean
)


print(
    "\nFeatures standardized successfully."
)


# ============================================================
# 13. PCA
# ============================================================

print("\n" + "=" * 75)
print("PRINCIPAL COMPONENT ANALYSIS")
print("=" * 75)


pca_full = PCA()


X_pca_full = pca_full.fit_transform(
    X_scaled
)


explained_variance = (
    pca_full
    .explained_variance_ratio_
)


cumulative_variance = np.cumsum(
    explained_variance
)


print(
    "\nVariance explained by each component:"
)


for i, value in enumerate(
    explained_variance
):

    print(

        f"PC{i + 1}: "
        f"{value:.4f} "
        f"({value * 100:.2f}%)"

    )


# ============================================================
# 14. CUMULATIVE EXPLAINED VARIANCE
# ============================================================

print(
    "\nCumulative variance:"
)


for i, value in enumerate(
    cumulative_variance
):

    print(

        f"PC1-PC{i + 1}: "
        f"{value:.4f} "
        f"({value * 100:.2f}%)"

    )


# ============================================================
# 15. SCREE PLOT
# ============================================================

plt.figure(
    figsize=(9, 5)
)


components = np.arange(
    1,
    len(explained_variance) + 1
)


plt.plot(

    components,

    explained_variance,

    "o-"

)


plt.xlabel(
    "Principal Component"
)

plt.ylabel(
    "Explained Variance Ratio"
)

plt.title(
    "PCA Scree Plot"
)

plt.xticks(
    components
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 16. CHOOSE NUMBER OF PCA COMPONENTS
# ============================================================

# We need at least 2 components for visualization

n_components_2 = min(
    2,
    X_scaled.shape[1]
)


pca_2 = PCA(
    n_components=n_components_2
)


Xp = pca_2.fit_transform(
    X_scaled
)


print(
    "\nPCA components used for clustering:",
    n_components_2
)


print(
    "Variance retained:",
    round(
        pca_2.explained_variance_ratio_.sum(),
        4
    )
)


# ============================================================
# 17. SILHOUETTE ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("FINDING OPTIMAL NUMBER OF CLUSTERS")
print("=" * 75)


# We need enough samples for clustering

if len(Xp) < 4:

    raise ValueError(
        "At least 4 rows are required "
        "for clustering."
    )


max_k = min(
    7,
    len(Xp) - 1
)


k_values = list(
    range(
        2,
        max_k + 1
    )
)


silhouette_scores = []


for k_value in k_values:

    model = KMeans(

        n_clusters=k_value,

        n_init=10,

        random_state=0
    )


    labels = model.fit_predict(
        Xp
    )


    score = silhouette_score(
        Xp,
        labels
    )


    silhouette_scores.append(
        score
    )


    print(

        f"k = {k_value} "
        f"| silhouette = "
        f"{score:.4f}"

    )


# ============================================================
# 18. SELECT BEST K
# ============================================================

best_index = int(
    np.argmax(
        silhouette_scores
    )
)


best_k = k_values[
    best_index
]


best_silhouette = silhouette_scores[
    best_index
]


print(
    "\nBest number of clusters:",
    best_k
)


print(
    "Best silhouette score:",
    round(
        best_silhouette,
        4
    )
)


# ============================================================
# 19. SILHOUETTE PLOT
# ============================================================

plt.figure(
    figsize=(9, 5)
)


plt.plot(

    k_values,

    silhouette_scores,

    "o-"

)


plt.axvline(

    best_k,

    linestyle="--",

    label=f"Best k = {best_k}"

)


plt.xlabel(
    "Number of Clusters (k)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "Choosing Number of Clusters"
)

plt.xticks(
    k_values
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 20. FINAL K-MEANS
# ============================================================

print("\n" + "=" * 75)
print("K-MEANS CLUSTERING")
print("=" * 75)


km = KMeans(

    n_clusters=best_k,

    n_init=10,

    random_state=0
)


labels = km.fit_predict(
    Xp
)


data["segment"] = labels


print(
    "\nSegments =",
    best_k
)


print(
    "Silhouette =",
    round(
        best_silhouette,
        4
    )
)


# ============================================================
# 21. PCA CLUSTER VISUALIZATION
# ============================================================

plt.figure(
    figsize=(10, 7)
)


plt.scatter(

    Xp[:, 0],

    Xp[:, 1],

    c=labels,

    cmap="viridis",

    s=30,

    alpha=0.75

)


# Plot cluster centers

centers = km.cluster_centers_


plt.scatter(

    centers[:, 0],

    centers[:, 1],

    marker="X",

    s=200,

    edgecolor="black",

    linewidth=1.5,

    label="Cluster center"

)


plt.xlabel(
    "PC1"
)

plt.ylabel(
    "PC2"
)

plt.title(
    "K-Means Clusters in PCA Space"
)

plt.legend()

plt.grid(
    alpha=0.2
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. PCA BIPLOT
# ============================================================

print("\n" + "=" * 75)
print("PCA BIPLOT")
print("=" * 75)


plt.figure(
    figsize=(11, 8)
)


plt.scatter(

    Xp[:, 0],

    Xp[:, 1],

    c=labels,

    cmap="viridis",

    s=25,

    alpha=0.65

)


# ------------------------------------------------------------
# Feature loadings
# ------------------------------------------------------------

loadings = pca_2.components_.T


# Scale arrows according to plot range

x_range = (
    Xp[:, 0].max()
    -
    Xp[:, 0].min()
)


y_range = (
    Xp[:, 1].max()
    -
    Xp[:, 1].min()
)


arrow_scale = min(
    x_range,
    y_range
) * 0.35


for i, feature in enumerate(
    numeric_features
):

    x = (
        loadings[i, 0]
        *
        arrow_scale
    )


    y_arrow = (
        loadings[i, 1]
        *
        arrow_scale
    )


    plt.arrow(

        0,

        0,

        x,

        y_arrow,

        color="black",

        alpha=0.7,

        head_width=0.08,

        length_includes_head=True

    )


    plt.text(

        x * 1.08,

        y_arrow * 1.08,

        feature,

        fontsize=9

    )


plt.axhline(
    0,
    linewidth=0.8,
    alpha=0.4
)

plt.axvline(
    0,
    linewidth=0.8,
    alpha=0.4
)


plt.xlabel(
    "PC1"
)

plt.ylabel(
    "PC2"
)

plt.title(
    "PCA Biplot with K-Means Segments"
)

plt.grid(
    alpha=0.2
)

plt.tight_layout()

plt.show()


# ============================================================
# 23. CLUSTER PROFILE
# ============================================================

print("\n" + "=" * 75)
print("CLUSTER PROFILES")
print("=" * 75)


profile = (

    data

    .groupby("segment")[numeric_features]

    .mean()

    .round(2)

)


profile["samples"] = (

    data

    ["segment"]

    .value_counts()

    .sort_index()

)


print(
    "\nCluster profile:"
)


print(
    profile
)


# ============================================================
# 24. STANDARDIZED CLUSTER PROFILE
# ============================================================

print("\n" + "=" * 75)
print("STANDARDIZED CLUSTER PROFILE")
print("=" * 75)


# Calculate mean standardized feature values
# for every cluster

standardized_df = pd.DataFrame(

    X_scaled,

    columns=numeric_features

)


standardized_df["segment"] = labels


standard_profile = (

    standardized_df

    .groupby("segment")[numeric_features]

    .mean()

    .round(2)

)


print(
    "\nValues:"
)

print(
    standard_profile
)


# ============================================================
# 25. HEATMAP OF CLUSTER PROFILES
# ============================================================

if best_k <= 15 and len(numeric_features) <= 30:

    plt.figure(

        figsize=(12, 6)

    )


    sns.heatmap(

        standard_profile,

        annot=True,

        fmt=".2f",

        cmap="coolwarm",

        center=0

    )


    plt.title(
        "Standardized Cluster Profiles"
    )


    plt.xlabel(
        "Features"
    )

    plt.ylabel(
        "Cluster"
    )


    plt.tight_layout()

    plt.show()


# ============================================================
# 26. AUTOMATIC CLUSTER INTERPRETATION
# ============================================================

print("\n" + "=" * 75)
print("CLUSTER INTERPRETATION")
print("=" * 75)


for cluster_id in sorted(
    data["segment"].unique()
):

    row = profile.loc[
        cluster_id
    ]


    sample_count = int(
        row["samples"]
    )


    print(
        f"\nCluster {cluster_id}:"
    )


    print(
        f"  Samples: {sample_count}"
    )


    # --------------------------------------------------------
    # Find strongest above-average features
    # --------------------------------------------------------

    zrow = standard_profile.loc[
        cluster_id
    ]


    sorted_features = (

        zrow

        .sort_values(
            ascending=False
        )

    )


    high_features = (

        sorted_features

        .head(
            min(
                3,
                len(sorted_features)
            )
        )

    )


    low_features = (

        sorted_features

        .tail(
            min(
                3,
                len(sorted_features)
            )
        )

    )


    print(
        "  Highest relative features:"
    )


    for feature, value in high_features.items():

        print(
            f"    {feature}: "
            f"{value:+.2f} SD"
        )


    print(
        "  Lowest relative features:"
    )


    for feature, value in low_features.items():

        print(
            f"    {feature}: "
            f"{value:+.2f} SD"
        )


# ============================================================
# 27. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("FINAL SUMMARY")
print("=" * 75)


print(
    "\nDataset:",
    file_path
)


print(
    "Rows:",
    len(data)
)


print(
    "Original numeric features:",
    len(numeric_features)
)


print(
    "Features used:",
    numeric_features
)


print(
    "\nNumber of clusters:",
    best_k
)


print(
    "Silhouette score:",
    round(
        best_silhouette,
        4
    )
)


print(
    "\nPCA variance retained by PC1 + PC2:",
    round(
        pca_2.explained_variance_ratio_.sum()
        * 100,
        2
    ),
    "%"
)
''

print(
    "\nCluster sizes:"
)


print(
    data["segment"]
    .value_counts()
    .sort_index()
)


print("\n")
print("=" * 75)
print("EXPERIMENT COMPLETED SUCCESSFULLY!")
print("=" * 75)
```
